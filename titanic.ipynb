```python
import pandas as pd #most of the cmd we have done, uses pandas, head, info, describe, average etc..
import seaborn as sns # visualisation lib, this contains our titanic dataset. makes beautiful graphs.
import matplotlib.pyplot as plt#we get graphs from seaborn, sure, but matplot helps us customize the very same graphs, custom x,y axis etc..

df = sns.load_dataset('titanic')
```

```python
#making graphs next...
df.groupby('sex')['survived'].sum().plot(kind='bar') #seaborn lib used to make graph, to customize, we see next lines of code..
plt.title('Survival by gender')
plt.ylabel('Number of survivors')
plt.show()
```

```python
df['age'].plot(kind='hist', bins=20) #age distribution histogram.
plt.title('Age distribution of passengers')
plt.xlabel('Age')
plt.show()
```
